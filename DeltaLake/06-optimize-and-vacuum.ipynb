```sql
CREATE OR REPLACE TABLE transactions (
  txn_id      INT,
  customer_id INT,
  amount      DOUBLE,
  category    STRING,
  txn_date    DATE
)
USING DELTA
COMMENT 'Transactions table — used to demonstrate OPTIMIZE and VACUUM'
TBLPROPERTIES (
  'delta.autoOptimize.optimizeWrite' = 'false',
  'delta.autoOptimize.autoCompact' = 'false'
);
```
